```python
%%writefile diagnostics.py

def monitor(func):
    def wrapper(value):
        print("Checking:", value)
        return func(value)
    return wrapper


@monitor
def check_reading(value):
    if not isinstance(value, (int, float)):
        raise ValueError("Invalid telemetry reading")

    if value >= 20:
        return "ABNORMAL"
    return "NORMAL"


def trace_abnormal(n):
    if n <= 0:
        return "Trace complete"
    return trace_abnormal(n - 1)
    
```

```python
from telemetry import generate_telemetry
from diagnostics import check_reading, trace_abnormal

transform = lambda x: x * 2

processed = 0
valid = 0
invalid = 0
abnormal = 0

for reading in generate_telemetry(LAST_NAME, SEED_NUM, FAVORITE_ARTIST):
    processed += 1

    try:
        reading = transform(reading)
        result = check_reading(reading)

        valid += 1

        if result == "ABNORMAL":
            abnormal += 1

    except ValueError:
        invalid += 1

if abnormal > 0:
    recursive_result = trace_abnormal(abnormal)
else:
    recursive_result = "No abnormal condition"

print("\n=== FINAL DIAGNOSTIC REPORT ===")
print("Processed readings:", processed)
print("Valid readings:", valid)
print("Invalid readings:", invalid)
print("Abnormal conditions:", abnormal)
print("Recursive analysis:", recursive_result)
```
