```python
LAST_NAME = '{PARRAS}'
SEED_NUM = 7
FAVORITE_ARTIST = '{DAUGHTRY}'


def generate_fault_code():
    return len(LAST_NAME) + SEED_NUM + len(FAVORITE_ARTIST)
```

```python

def trace_fault(code):
    print("Fault Code:", code)

    if code <= 0:
        return

    trace_fault(code - 1)
    
    
```

```python
fault_code = generate_fault_code()

print("Starting Fault Trace:")
trace_fault(fault_code)
```
